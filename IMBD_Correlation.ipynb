Data Exploration

```python
#Check for missing data values

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col,pct_missing))
```

```python

# 1. Handle Categorical
df['Certificate'] = df['Certificate'].fillna('Unknown')

# 2. Handle Numerical (Score)
df['Meta_score'] = df['Meta_score'].fillna(df['Meta_score'].median())

# 3. Handle Numerical (Financial)
# Convert everything to string first
df['Gross'] = df['Gross'].astype(str).str.replace(',', '')

# Clean up ALL possible variations of text-based missing value markers
df['Gross'] = df['Gross'].replace(['nan', 'NaN', '<NA>', '<na>', ''], '0')

# to convert to standard integer
df['Gross'] = df['Gross'].astype('int64')
```

```python
df.dtypes
```

```python
#Chnage the data type of the type of column 

df['No_of_Votes'] = df['No_of_Votes'].astype('float')
```

```python
df.No_of_Votes.dtypes
```

```python
#The data type is Integer

df['No_of_Votes'] = df['No_of_Votes'].astype('int')
```

```python
df.No_of_Votes.dtypes
```

```python
df.head()
```

```python
#Sort the values as per Gross 

df.sort_values(by=['Gross'], inplace = False, ascending= False).head()
```
